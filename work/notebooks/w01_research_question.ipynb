# Research Question

## Decision

Which content pages should be prioritized for refresh in order to maximize future search performance?

## Unit of Analysis

Each individual content page.

## Output

A ranked list of content pages based on their estimated opportunity for improvement using historical search, engagement, and content characteristics.
